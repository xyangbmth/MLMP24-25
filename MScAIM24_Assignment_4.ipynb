## Assignment 4: Comparison of Mean and Median Filter   (20 marks)

Image denoising is an essential task in image processing, with common methods including mean filtering and median filtering. Mean filtering smooths an image by averaging pixel values within a neighborhood, while median filtering replaces a pixel's value with the median of its neighborhood. These methods perform differently when dealing with various types of noise.

**Task Description**

1. Write a Python program to apply mean filtering and median filtering to a noisy grayscale image (3×3 filter window size).
2. Display the denoised results for both filters side by side with the original noisy image for visual comparison.
3. Provide a brief comparative analysis of the two methods.


### Task 1. Mean filtering (8 marks)

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
test_img = cv2.imread("Test_noisy_img.png",0)
plt.imshow(test_img,cmap="gray")
test_img.shape
```

```python
Pixel_list1 = []
for i in range(1,255):
    for j in range(1,255):

        neighbourlist=[]


        neighbourlist.append(test_img[i-1][j-1])
        neighbourlist.append(test_img[i-1][j])
        neighbourlist.append(test_img[i-1][j+1])
        neighbourlist.append(test_img[i][j-1])
        neighbourlist.append(test_img[i][j])
        neighbourlist.append(test_img[i][j+1])
        neighbourlist.append(test_img[i+1][j-1])
        neighbourlist.append(test_img[i+1][j])
        neighbourlist.append(test_img[i+1][j+1])
        
        Pixel_list1.append(np.mean(neighbourlist))

Conv_image1 = np.array(Pixel_list1).reshape(254,254)
plt.imshow(Conv_image1,cmap="gray",vmin=0,vmax=255)
Conv_image1.shape
```

### Task 2. Median filtering (8 marks)

```python
Pixel_list2 = []
for i in range(1,255):
    for j in range(1,255):

        neighbourlist=[]


        neighbourlist.append(test_img[i-1][j-1])
        neighbourlist.append(test_img[i-1][j])
        neighbourlist.append(test_img[i-1][j+1])
        neighbourlist.append(test_img[i][j-1])
        neighbourlist.append(test_img[i][j])
        neighbourlist.append(test_img[i][j+1])
        neighbourlist.append(test_img[i+1][j-1])
        neighbourlist.append(test_img[i+1][j])
        neighbourlist.append(test_img[i+1][j+1])
        
        Pixel_list2.append(np.median(neighbourlist))

Conv_image2 = np.array(Pixel_list2).reshape(254,254)
plt.imshow(Conv_image2,cmap="gray",vmin=0,vmax=255)
Conv_image2.shape
```

### Task 3. Plotting and Analyzing (4 marks)

```python
plt.subplot(1,2,1)
plt.title("Mean filtering")
plt.imshow(Conv_image1,cmap="gray",vmin=0,vmax=255)
plt.axis('off')

plt.subplot(1,2,2)
plt.title("Median filtering")
plt.imshow(Conv_image2,cmap="gray",vmin=0,vmax=255)
plt.axis('off')
```

Median filtering is especially good at removing "salt-and-pepper noise" from images, which are those random bright or dark spots that pop up. Unlike average filtering, which smooths out everything by taking the mean of the surrounding pixels, median filtering replaces a pixel's value with the median of its neighbors. This works better because the median isn't affected by extremely high or low values—so it keeps the noise from messing up the result. Plus, it helps preserve the edges of objects in the image, which average filtering tends to blur. That's why median filtering is a go-to choice for cleaning up noisy images!
